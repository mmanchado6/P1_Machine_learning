```python
import joblib
import pandas as pd
from pathlib import Path
# Cargar modelo entrenado

modelo = joblib.load("modelo_final.pkl")

# Cargar dataset de competición
file_path = Path("content/P1") / "bank_competition.pkl"
df_test = pd.read_pickle(file_path)

# Generar predicciones
predicciones = modelo.predict(df_test)

# Crear DataFrame de salida
df_pred = pd.DataFrame({
    "target": predicciones
})

# Guardar CSV
df_pred.to_csv("predicciones.csv", index=False)

print("Archivo 'predicciones.csv' generado correctamente")
```
